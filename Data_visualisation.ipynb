```python
def plot_time_series(prefix='D', cols=None, display_ct=32):

    if cols is None:
        COLS = df.columns[2:-1]
        COLS = np.sort( [int(x[2:]) for x in COLS if x[0]==prefix] )
        print(COLS)
        COLS = [f'{prefix}_{x}' for x in COLS]
        print('#'*25)
        print(f'Plotting all {len(COLS)} columns with prefix {prefix}')
        print('#'*25)
    else:
        COLS = [f'{prefix}_{x}' for x in cols]
        print('#'*25)
        print(f'Plotting {len(COLS)} columns with prefix {prefix}')
        print('#'*25)

    for c in COLS:
        tmp = df[['customer_ID','S_2',c,'target']].copy()
        tmp2 = tmp.groupby(['customer_ID','target'])[['S_2',c]].agg(list).reset_index()
        tmp3 = tmp2.loc[tmp2.target==1]
        tmp4 = tmp2.loc[tmp2.target==0]

        spec = gridspec.GridSpec(ncols=2, nrows=1,
                             width_ratios=[3, 1], wspace=0.1,
                             hspace=0.5, height_ratios=[1])
        fig = plt.figure(figsize=(20,10))
        ax0 = fig.add_subplot(spec[0])

        # PLOT 32 DEFAULT CUSTOMERS AND 32 NON-DEFAULT CUSTOMERS
        t0 = []; t1 = []
        for k in range(display_ct):
            try:
                # PLOT DEFAULTING CUSTOMERS
                row = tmp3.iloc[k]
                ax0.plot(row.S_2,row[c],'-o',color='blue')
                t1 += row[c]
                # PLOT NON-DEFAULT CUSTOMERS
                row = tmp4.iloc[k]
                ax0.plot(row.S_2,row[c],'-o',color='red')
                t0 += row[c]
            except:
                pass
        plt.title(f'Feature {c} (Key: BLUE=DEFAULT, orange=no default)',size=18)

        # PLOT HISTOGRAMS
        ax1 = fig.add_subplot(spec[1])
        try:
            # COMPUTE BINS
            t = t0+t1; mn = np.nanmin(t); mx = np.nanmax(t)
            if mx==mn:
                mx += 0.01; mn -= 0.01
            bins = np.arange(mn,mx+(mx-mn)/20,(mx-mn)/20 )
            # PLOT HISTOGRAMS
            if np.sum(np.isnan(t1))!=len(t1):
                ax1.hist(t1,bins=bins,orientation="horizontal",alpha = 0.8,color='blue')
            if np.sum(np.isnan(t0))!=len(t0):
                ax1.hist(t0,bins=bins,orientation="horizontal",alpha = 0.8,color='red')
        except:
            pass
        plt.show()
```

```python
import random

def random_col_selection(prefix, catCols, numberOfCol):
    # numberOfCol - Setting upper bound for number of col selection
    cols = np.sort( [x for x in df.columns[2:-1] if x[0]==prefix] )
    randomSelectedCols = []
    if len(cols) < numberOfCol:
        print("Randomly selected column count has to be less than total number of cols")
        return
    for col in cols:
        if len(randomSelectedCols) >= numberOfCol:
            break
        if random.randint(0, 1) and col not in catCols:
            randomSelectedCols.append(col[2:])

    return randomSelectedCols
```

```python
#performing boxplot
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import calendar
from scipy.stats import zscore

# Load the data with headers to preserve original column names
#df = pd.read_csv("data/train_data.csv", skiprows=20000, nrows=20000, header=0)

# Create a copy to avoid altering original DataFrame
df_copy = df.copy()

# Convert assumed date column 'S_2' to datetime in the copy
df_copy['date'] = pd.to_datetime(df_copy['S_2'], errors='coerce')

# Extract month and day for grouping in the copy
df_copy['month'] = df_copy['date'].dt.month
df_copy['day'] = df_copy['date'].dt.day

# Load labels and merge to include 'target'
df_ids = pd.read_csv("data/train_labels.csv", usecols=["customer_ID", "target"])
df_copy = df_copy.merge(df_ids, on='customer_ID', how='left')

# Filter defaulters for plotting
df_defaulters = df_copy[df_copy['target'] == 1]

# Rename feature column for clarity (replace 'B_1' with your feature)
feature_col = 'B_1'
df_defaulters = df_defaulters.rename(columns={feature_col: 'feature'})

# Example Outlier Detection Functions
def detect_outliers_iqr(data):
    Q1 = np.percentile(data, 25)
    Q3 = np.percentile(data, 75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return data[(data < lower_bound) | (data > upper_bound)]

def detect_outliers_zscore(data, threshold=2.5):
    if len(data) < 2:
        return pd.Series(dtype=data.dtype)
    zs = zscore(data)
    return data[(np.abs(zs) > threshold)]

outlier_detector = detect_outliers_zscore  # or detect_outliers_iqr

# Plotting boxplot with outliers highlighted
plt.figure(figsize=(22, 10))
ax = sns.boxplot(
    x='day',
    y='feature',
    hue='month',
    data=df_defaulters,
    showfliers=False,
    palette='tab20',
    dodge=True
)

width = 0.8
months = sorted(df_defaulters['month'].unique())
days = sorted(df_defaulters['day'].unique())
num_months = len(months)
width_per_month = width / num_months

all_outliers = []

for i, day in enumerate(days):
    for j, month in enumerate(months):
        group = df_defaulters[(df_defaulters['day'] == day) & (df_defaulters['month'] == month)]['feature']
        if group.empty:
            continue
        outliers = outlier_detector(group)
        if not outliers.empty:
            x = i - width/2 + width_per_month/2 + j * width_per_month
            jitter = np.random.uniform(-width_per_month/4, width_per_month/4, size=len(outliers))
            ax.scatter(x + jitter, outliers, color='red', s=30, edgecolor='black', alpha=0.7)
            all_outliers.append((month, day, outliers.values))

# Print summary of detected outliers:
if all_outliers:
    print(f"Detected outliers in total {len(all_outliers)} groups.")
    for month, day, values in all_outliers[:5]:
        print(f"Month: {calendar.month_name[month]}, Day: {day}, Outliers count: {len(values)}")
        print(f"Sample outlier values: {values[:5]}")
else:
    print("No outliers detected.")

plt.title(f'Boxplot of Feature {feature_col} by Day and Month for Defaulters\n(with multiple outliers detected)', fontsize=16)
plt.xlabel('Day of Month', fontsize=14)
plt.ylabel(f'Feature {feature_col}', fontsize=14)
plt.xticks(rotation=45)

ylim = ax.get_ylim()
ax.set_ylim(ylim[0] - 0.1 * abs(ylim[0]), ylim[1] + 0.1 * abs(ylim[1]))

handles, labels = ax.get_legend_handles_labels()
from collections import OrderedDict
unique_labels = OrderedDict(zip(labels, handles))
plt.legend(unique_labels.values(), unique_labels.keys(), title='Month', bbox_to_anchor=(1.05, 1),
           loc='upper left', fontsize=12)

plt.tight_layout()
plt.show()
```
